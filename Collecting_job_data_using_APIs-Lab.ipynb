#### Write a function to get the number of jobs for the Python technology.

 ##### The keys in the json are

You can also view  the json file contents  from the following <a href = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DA0321EN-SkillsNetwork/labs/module%201/Accessing%20Data%20Using%20APIs/jobs.json">json</a> URL.



```python
api_url="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DA0321EN-SkillsNetwork/labs/module%201/Accessing%20Data%20Using%20APIs/jobs.json"

def get_number_of_jobs_T(technology):
    response = requests.get(api_url)
    data = response.json()
    
    number_of_jobs = 0
    
    for job in data:
        if technology.lower() in job['Key Skills'].lower():
            number_of_jobs += 1
    return technology,number_of_jobs

```

Calling the function for Python and checking if it works.


```python
tech, count = get_number_of_jobs_T("Python")
print(f"Number of jobs for {tech}: {count}")
```

Collect the number of job postings for the following languages using the API:

```python
import requests
from openpyxl import Workbook

# API URL
api_url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DA0321EN-SkillsNetwork/labs/module%201/Accessing%20Data%20Using%20APIs/jobs.json"

# Function to get number of jobs for a technology
def get_number_of_jobs_T(technology):
    response = requests.get(api_url)
    data = response.json()
    
    number_of_jobs = 0
    for job in data:
        if technology.lower() in job['Key Skills'].lower():
            number_of_jobs += 1
            
    return technology, number_of_jobs

# List of technologies
technologies = [
    "C",
    "C#",
    "C++",
    "Java",
    "JavaScript",
    "Python",
    "Scala",
    "Oracle",
    "SQL Server",
    "MySQL Server",
    "PostgreSQL",
    "MongoDB"
]

# Create workbook and select active sheet
wb = Workbook()
ws = wb.active

# Write header row
ws.append(["Technology", "Number of Job Postings"])

# Loop through technologies and write results to Excel
for tech in technologies:
    tech_name, count = get_number_of_jobs_T(tech)
    ws.append([tech_name, count])

# Save workbook
wb.save("job-postings.xlsx")
```

```python
import requests  # to make GET and POST requests
from openpyxl import Workbook  # to create Excel files
```

```python
api_url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DA0321EN-SkillsNetwork/labs/module%201/Accessing%20Data%20Using%20APIs/jobs.json"

# GET request
r_get = requests.get(api_url)

# Check status
if r_get.status_code == 200:
    print("GET request successful!")
else:
    print("GET request failed:", r_get.status_code)

# Convert to JSON
data = r_get.json()
```

```python
post_url = "http://httpbin.org/post"
payload = {"technology": "Python", "location": "Los Angeles"}

# POST request
r_post = requests.post(post_url, data=payload)

# Check status
if r_post.status_code == 200:
    print("POST request successful!")
else:
    print("POST request failed:", r_post.status_code)
```

```python
print("GET URL:", r_get.url)
print("POST URL:", r_post.url)

print("\nGET request body (should be empty):", r_get.request.body)
print("POST request body (form data):", r_post.request.body)
print("POST form returned by server:", r_post.json()['form'])
```

```python
def get_number_of_jobs_T(technology):
    number_of_jobs = 0
    for job in data:
        if technology.lower() in job['Key Skills'].lower():
            number_of_jobs += 1
    return technology, number_of_jobs
```

```python
def get_number_of_jobs_Python_by_location(location):
    number_of_jobs = 0
    for job in data:
        if "Python" in job['Key Skills'] and location.lower() in job['Location'].lower():
            number_of_jobs += 1
    return location, number_of_jobs
```

```python
technologies = [
    "C", "C#", "C++", "Java", "JavaScript", "Python",
    "Scala", "Oracle", "SQL Server", "MySQL Server",
    "PostgreSQL", "MongoDB"
]
```

```python
wb = Workbook()
ws = wb.active

# Write header row
ws.append(["Technology / Location", "Number of Job Postings"])
```

```python
# Count jobs for all technologies
for tech in technologies:
    tech_name, count = get_number_of_jobs_T(tech)
    ws.append([tech_name, count])
```

```python
# Example locations
locations = ["Los Angeles", "New York", "San Francisco", "Washington DC", "Seattle", "Austin", "Detroit"]

for loc in locations:
    location, count = get_number_of_jobs_Python_by_location(loc)
    ws.append([f"Python in {location}", count])
```

```python
wb.save("github-job-postings.xlsx")
print("Excel file saved as 'github-job-postings.xlsx'")
```
