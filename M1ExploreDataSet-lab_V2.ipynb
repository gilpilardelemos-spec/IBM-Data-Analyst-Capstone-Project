```python
import pandas as pd
import numpy as np

# Example df
# df = pd.DataFrame({'age': [...]})

# Map age categories to numeric midpoints
age_mapping = {
    'Under 18 years old': 17/2,      # midpoint = 8.5
    '18-24 years old': (18+24)/2,    # midpoint = 21
    '25-34 years old': (25+34)/2,    # midpoint = 29.5
    '35-44 years old': (35+44)/2,    # midpoint = 39.5
    '45-54 years old': (45+54)/2,    # midpoint = 49.5
    '55-64 years old': (55+64)/2,    # midpoint = 59.5
    '65 years or older': 70,          # approximate midpoint
    'Prefer not to say': np.nan       # treat as missing
}

# Create a numeric column
df['age_numeric'] = df['age'].map(age_mapping)

# Compute the mean, ignoring NaN
mean_age = df['age_numeric'].mean()

print(f"The mean age of the participants is: {mean_age:.1f} years")
```
